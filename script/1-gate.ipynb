# Gate di Pauli

```python
from qiskit import QuantumCircuit
import matplotlib.pyplot as plt

# 1. Creazione dei circuiti, uno per ogni porta
circuiti = {}

# Identità
qc = QuantumCircuit(1)
qc.id(0)
circuiti["I"] = qc

# Pauli-X
qc = QuantumCircuit(1)
qc.x(0)
circuiti["X"] = qc

# Pauli-Y
qc = QuantumCircuit(1)
qc.y(0)
circuiti["Y"] = qc

# Pauli-Z
qc = QuantumCircuit(1)
qc.z(0)
circuiti["Z"] = qc

# Hadamard
qc = QuantumCircuit(1)
qc.h(0)
circuiti["H"] = qc

# C-NOT (qubit 0 = controllo, qubit 1 = bersaglio)
qc = QuantumCircuit(2)
qc.cx(0, 1)
circuiti["CNOT"] = qc

# 2. Rendering grafico e salvataggio
for nome, circuito in circuiti.items():
    fig = circuito.draw(output='mpl', style='iqp')
    fig.suptitle(f"Porta {nome}")
    # fig.savefig(f"porta_{nome}.png", dpi=300, bbox_inches='tight') # Salvataggio Foto
    display(fig)          # mostra l'immagine in Jupyter
    plt.close(fig)        # evita duplicati nell'output
```
