### Definizione del circuito di prova

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

# 1. Creazione del circuito
qc = QuantumCircuit(1, 1)
qc.h(0)
qc.measure(0, 0)

# 2. Rendering grafico per Jupyter
# Usando output='mpl' si attiva la grafica Matplotlib
qc.draw(output='mpl', style='iqp')
```

### Risultato su 1000 campioni di prova, simulati

```python
# Simulazione dell'esperimento
simulator = AerSimulator()
job = simulator.run(qc, shots=1000)
counts = job.result().get_counts()

# Istogramma ad alta risoluzione
plot_histogram(counts, color='#6200ee', title="Risultato della Misura del Qubit")
```
