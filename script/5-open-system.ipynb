# Open System Architecture

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from IPython.display import display

plt.rcParams["mathtext.fontset"] = "cm"

GATE_COLOR = "#9F1853"   # colore porte dello stile iqp
FS = 26                  # dimensione del testo
LW = 2.2                 # spessore dei fili


def wire(ax, x0, x1, y):
    ax.plot([x0, x1], [y, y], color="k", lw=LW, solid_capstyle="butt", zorder=1)


def gate(ax, x, y0, w, h, label=r"$U$"):
    ax.add_patch(Rectangle((x, y0), w, h, fc=GATE_COLOR, ec=GATE_COLOR, zorder=2))
    ax.text(x + w / 2, y0 + h / 2, label, color="white",
            ha="center", va="center", fontsize=FS + 4, zorder=3)


def new_fig(width, height, xlim, ylim):
    fig, ax = plt.subplots(figsize=(width, height))
    ax.set_xlim(*xlim)
    ax.set_ylim(*ylim)
    ax.set_aspect("equal")
    ax.axis("off")
    return fig, ax


# ---------- Figura 1: sistema chiuso ----------
fig1, ax = new_fig(7, 2.2, (0, 10.5), (-1.3, 1.3))
ax.text(1.2, 0, r"$\rho$", ha="right", va="center", fontsize=FS)
wire(ax, 1.5, 3.3, 0)
gate(ax, 3.3, -0.75, 1.5, 1.5)
wire(ax, 4.8, 6.6, 0)
ax.text(6.9, 0, r"$U\rho U^\dagger$", ha="left", va="center", fontsize=FS)

fig1.savefig("sistema_chiuso.png", dpi=300, bbox_inches="tight", transparent=False)
display(fig1)
plt.close(fig1)


# ---------- Figura 2: sistema aperto ----------
fig2, ax = new_fig(7.5, 3.2, (0, 11.5), (-2.0, 2.0))
y_sys, y_env = 0.9, -0.9
ax.text(2.4, y_sys, r"$\rho$", ha="right", va="center", fontsize=FS)
ax.text(2.4, y_env, r"$\rho_{\mathrm{env}}$", ha="right", va="center", fontsize=FS)
wire(ax, 2.7, 4.3, y_sys)
wire(ax, 2.7, 4.3, y_env)
gate(ax, 4.3, -1.5, 1.8, 3.0)
wire(ax, 6.1, 7.7, y_sys)
wire(ax, 6.1, 7.7, y_env)
ax.text(8.0, y_sys, r"$\mathcal{E}(\rho)$", ha="left", va="center", fontsize=FS)

fig2.savefig("sistema_aperto.png", dpi=300, bbox_inches="tight", transparent=False)
display(fig2)
plt.close(fig2)
```
